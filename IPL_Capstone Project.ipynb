**10 Best Bowling Figures**

```python
df.head()
```

```python
df['high_wickets'] =df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
```

```python
df.head()
```

```python
df['high_wickets'] =df['high_wickets'].astype(int)
```

```python
data = df.groupby('best_bowling')['high_wickets'].sum().sort_values(ascending = False).head(10)
data
```

```python
sns.barplot(x = data.values, y = data.index,palette = 'viridis')
```
