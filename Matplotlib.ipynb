```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,12)
x
```

```python
y = np.linspace(0,10,12)
y
```

```python
plt.plot(x,y)
plt.title('Graph')
plt.xlabel('x_axis')
plt.ylabel('y_axis')
```

```python
# SUBPLOT:
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x**2,x**5)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x**7,y**3)
```

```python
#OBJECT ORIENTED WAY:
```

```python
fig = plt.figure()
axis = fig.add_axes([0.1,0.1,.8,.8])
plt.plot(x,y)
axis2  = fig.add_axes([0.5,0.2,.3,.2])
plt.plot(x,y)
```

```python
#TYPES OF PLOTS:
```

```python
plt.scatter(x,y)
```

```python
plt.hist(x, bins=10, edgecolor='black')
plt.show()
```

```python
plt.boxplot(x)
```

```python
plt.boxplot(y)
```

```python

#SAVING PLOT:
```

```python
fig = plt.figure()
axis = fig.add_axes([0.1,0.1,.8,.8])
plt.plot(x,y)
plt.savefig('Bacis_plot.png')
```
