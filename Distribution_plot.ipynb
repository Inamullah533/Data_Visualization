```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('planets')
df
```

```python
df['year'].unique()
```

```python
plt.subplot(1,2,1)
sns.histplot(df['mass'],kde = True)
plt.subplot(1,2,2)
sns.histplot(df['year'],kde = True)
```

```python
sns.jointplot(x = 'mass', y = 'year' , data = df , kind = 'scatter')
```

```python
sns.jointplot(x = 'mass', y = 'year' , data = df , kind = 'reg')
```

```python
sns.pairplot(df,hue = 'year', palette = 'rainbow')
```
