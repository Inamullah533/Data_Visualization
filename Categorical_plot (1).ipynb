```python
import matplotlib.pyplot as plt
import seaborn as sns
```

**#Categorical Data plot :
boxplot
countplot
barplot
voilinplot
stripplot
swarmplot**

```python
df = sns.load_dataset('tips')
df
```

**countplot**

```python
sns.countplot(x = df['sex'] , hue = df['smoker'] )
```

**Barplot**

```python
import numpy as np
sns.barplot(x=df['day'], y=df['total_bill'], estimator=np.sum)
```

**Boxplot**

```python
sns.boxplot(x = df['day'], y= df['total_bill'],palette = 'rainbow')
```

**Violinplot**

```python
sns.violinplot(x = df['day'], y= df['total_bill'] , data = df)
```

**Stripplot**

```python
sns.stripplot(x = df['day'], y= df['total_bill'] , data = df)
```

**Swarmplot**

```python
sns.swarmplot(x = df['day'], y= df['total_bill'] , data = df)
```
